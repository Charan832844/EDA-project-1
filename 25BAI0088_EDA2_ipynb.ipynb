```python
import pandas as pd
```

```python
df1AS =  pd.DataFrame({ 'Jersey Number': [7, 10, 18, 45, 33, 11, 17, 8, 1, 42, 48],
                       'Average Score' : [98, 66, 31, 51, 28, 56, 48, 32, 52, 73, 92]})
df2AS =  pd.DataFrame({'Jersey Number': [2, 4, 6, 28, 10, 12, 14, 16, 38, 20],
                       'Average Score': [98, 93, 44, 77, 69, 56, 31, 53, 78, 93]})

df1HD =  pd.DataFrame({ 'Jersey Number': [7, 10, 18, 45, 33, 11, 17, 8, 1, 42, 48],
                       'Hundreds' : [39, 49, 55, 77, 52, 86, 41, 77, 73, 51, 86]})
df2HD =  pd.DataFrame({'Jersey Number': [2, 4, 6, 28, 10, 12, 14, 16, 38, 20],
                       'Hundreds': [93, 44, 78, 97, 87, 89, 39, 43, 88, 78]})
```

```python
dfSE = pd.concat([df1AS, df2AS], ignore_index=True)
dfML = pd.concat([df1HD, df2HD], ignore_index=True)


df = pd.concat([dfML, dfSE], axis=1)
df
```

```python
dfSE = pd.concat([df1AS, df2AS], ignore_index=True)
dfML = pd.concat([df1HD, df2HD], ignore_index=True)

df = dfSE.merge(dfML, how='inner')
df
```

```python
dfSE = pd.concat([df1AS, df2AS], ignore_index=True)
dfML = pd.concat([df1HD, df2HD], ignore_index=True)

df = dfSE.merge(dfML, how='left')
df
```

```python
dfSE = pd.concat([df1AS, df2AS], ignore_index=True)
dfML = pd.concat([df1HD, df2HD], ignore_index=True)

df = dfSE.merge(dfML, how='right')
df
```
